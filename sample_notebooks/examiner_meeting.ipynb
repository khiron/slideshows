```python
import cogent3
import madb 

seqs = {
          #       X
    "seq1": "ACGACCGCAT",
    "seq2": "ACGACTGCAT"
}
dbg = madb.DeBruijnGraph(kmer_length=3, moltype=cogent3.DNA)
dbg.add_sequence(seqs)
madb.utils.mm(dbg.to_mermaid(highlight_CSS="color: red"))
```

```python
for bubble in dbg.bubbles():
    print(f'Bubble: {bubble}')
for braid in dbg.braids():
    print(f'Braid: {braid}')
```

```python
bubbles = dbg.bubbles()
bubbles.minimize_bubbles()  # reject bubbles that meet Karlin Statistic test 
for bubble in bubbles:
    print(f'Bubble: {bubble}')
for braid in madb.Braids(dbg, bubbles): # given the new set of bubbles we can now find the braids
    print(f'Braid: {braid}')
```
