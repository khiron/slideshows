```python
import cogent3
import os
from cogent3.evolve.models import JC69
from cogent3.app.composable import NotCompleted
from cogent3.app.io import write_seqs

def to_taxonomic_name(seqname):
    return seqname.split("-")[0].replace("_", " ")

mammals_dir = '../mammals'  
hcm_dir = '../hcm'

# Use the modern cogent3 app interface for progressive aligner
aligner = cogent3.get_app("progressive_align", model="nucleotide")

for filename in os.listdir('../mammals'):
    if filename.endswith('.fa'):
        print(f'Loading {filename}')
        # Load unaligned sequences
        unaligned = cogent3.load_unaligned_seqs(filename, moltype=cogent3.DNA)
        unaligned = unaligned.rename_seqs(to_taxonomic_name)

        try:
            # Take human, chimp, and mouse sequences
            subset = unaligned.take_seqs(["homo sapiens", "pan troglodytes", "mus musculus"])  # 
        except KeyError as e:
            print(f"Skipping {filename}: missing required species — {e}")
            continue

        # Align with progressive aligner
        result = aligner(subset)

        if isinstance(result, NotCompleted):
            print(f"Alignment failed for {filename}: {result}")
            continue


        # Save aligned result
        wrapped = cogent3.Alignment(result.to_dict(), moltype="dna").omit_gap_pos()
        output_filename = f'HMP_{filename}'
        print(f'Saving to {output_filename}')
        with open(output_filename, "w") as f:
            f.write(result.to_fasta())

```
